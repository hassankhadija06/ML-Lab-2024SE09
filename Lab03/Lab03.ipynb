### Task 2(h) Analysis: What Happens and Why

- **What happens:**
  Evaluating `df[df["Price"] > 1_000_000]` raises a `TypeError` (e.g., `TypeError: Invalid comparison between dtype=str and int` or `TypeError: '>' not supported between instances of 'str' and 'int'`).

- **Why it happens:**
  In the raw, uncleaned dataset, the `Price` column is stored as an **object** (text/string) data type rather than a numeric data type (`int64`). This is because multiple entries were entered with currency symbols, spaces, and commas (such as `"Rs 1,430,000"`). In Python and Pandas, comparison operators (`>`, `<`, etc.) are not defined between text strings and numbers. Even a single string in a column forces pandas to treat the entire column as text, preventing direct mathematical filtering. To perform comparisons and quantitative calculations, the column must first be cleaned (stripping `"Rs"` and commas) and cast to integer (`int64`), which is executed in Task 3.

## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

```python
# T3: your code here
# (a) Missing values: count per column, drop missing Brand/Model, fill text with "Unknown"
print("T3a: Missing values per column before cleaning:")
print(df.isna().sum())

initial_count = len(df)
df = df.dropna(subset=["Brand", "Model"])
rows_dropped_missing = initial_count - len(df)
print(f"\nT3a: Rows removed with missing Brand or Model: {rows_dropped_missing}")

df["Registered City"] = df["Registered City"].fillna("Unknown")
df["Fuel"] = df["Fuel"].fillna("Unknown")
df["Transaction Type"] = df["Transaction Type"].fillna("Unknown")
# Leave Year and KMs Driven as NaN

# (b) Formatting: city names strip/title, Price remove "Rs" and commas
df["Registered City"] = df["Registered City"].str.strip().str.title()
df["Price"] = df["Price"].str.replace("Rs", "", regex=False).str.replace(",", "", regex=False).str.strip()
print("\nT3b: Standardized city text formatting and stripped 'Rs' & commas from Price.")

# (c) Data types: Price to int64, Year to Int64
df["Price"] = df["Price"].astype("int64")
df["Year"] = df["Year"].astype("Int64")
print("\nT3c: Converted column data types:")
print(df[["Price", "Year"]].dtypes)

# (d) Inconsistent categories: Sukkar -> Sukkur, Corrolla Altis -> Corolla Altis
cities_before = df["Registered City"].nunique()
df["Registered City"] = df["Registered City"].replace({"Sukkar": "Sukkur"})
df["Model"] = df["Model"].replace({"Corrolla Altis": "Corolla Altis"})
cities_after = df["Registered City"].nunique()
print(f"\nT3d: Number of different cities before: {cities_before}, after replacement: {cities_after}")

# (e) Duplicates: count exact duplicate ads, drop them, print new row count
dup_count = df.duplicated().sum()
print(f"\nT3e: Exact duplicate ads count: {dup_count}")
df = df.drop_duplicates()
print(f"T3e: New row count after dropping duplicates: {len(df)}")

# (f) Invalid values:
# set KMs Driven to NaN when < 100, > 1,000,000, or in [12345, 123456]
km_invalid = (df["KMs Driven"] < 100) | (df["KMs Driven"] > 1_000_000) | (df["KMs Driven"].isin([12345, 123456]))
print(f"\nT3f: Invalid KMs Driven values replaced with NaN: {km_invalid.sum()}")
df.loc[km_invalid, "KMs Driven"] = np.nan

# Drop fake prices and Suzuki/Daihatsu > 5M
fake_prices = df["Price"].isin([12345678, 87654321])
expensive_budget = (df["Brand"].isin(["Suzuki", "Daihatsu"])) & (df["Price"] > 5_000_000)
invalid_price_rows = fake_prices | expensive_budget
print(f"T3f: Dropped invalid price ads: {invalid_price_rows.sum()}")
df = df[~invalid_price_rows]

# (g) Self-check: final shape must be (20484, 9)
print(f"\nT3g: Final shape after cleaning: {df.shape}")
assert df.shape == (20484, 9), f"Self-check FAILED: Expected (20484, 9), got {df.shape}"
print("T3g: Self-check SUCCESS: Shape matches (20484, 9) exactly!")
```

## Task 4 · Analyse
- **(a)** New column `Price_lakh` = price in lakh rupees (÷ 100,000).
- **(b)** New column `Age` = 2020 − `Year`.
- **(c)** The 5 most common brands.
- **(d)** The 5 cities with the most ads.
- **(e)** Median `Price_lakh` for each fuel type, highest first.
- **(f)** Median `Price_lakh` of a **Corolla GLI** for each year 2010–2018. How much value does it lose per year?
- **(g)** The 5 most expensive ads. Does one of them look wrong? Why?
- **(h)** Number of ads and median price for Cash vs Installment/Leasing. Interpret in 1–2 lines.

```python
# T4: your code here
# (a) New column Price_lakh
df["Price_lakh"] = df["Price"] / 100_000
print("T4a: First 5 rows with Price_lakh:")
print(df[["Brand", "Model", "Price", "Price_lakh"]].head(5))

# (b) New column Age = 2020 - Year
df["Age"] = 2020 - df["Year"]
print("\nT4b: First 5 rows with Age:")
print(df[["Brand", "Model", "Year", "Age"]].head(5))

# (c) The 5 most common brands
top5_brands = df["Brand"].value_counts().head(5)
print("\nT4c: 5 most common brands:")
print(top5_brands)

# (d) The 5 cities with the most ads
top5_cities = df["Registered City"].value_counts().head(5)
print("\nT4d: 5 cities with the most ads:")
print(top5_cities)

top5_known_cities = df[df["Registered City"] != "Unknown"]["Registered City"].value_counts().head(5)
print("\nT4d (excluding 'Unknown'): 5 cities with the most ads:")
print(top5_known_cities)

# (e) Median Price_lakh for each fuel type, highest first
median_fuel = df.groupby("Fuel")["Price_lakh"].median().sort_values(ascending=False)
print("\nT4e: Median Price_lakh for each fuel type (highest first):")
print(median_fuel)

# (f) Median Price_lakh of Corolla GLI for each year 2010–2018
gli_ads = df[(df["Model"] == "Corolla GLI") & (df["Year"] >= 2010) & (df["Year"] <= 2018)]
gli_median = gli_ads.groupby("Year")["Price_lakh"].median()
print("\nT4f: Median Price_lakh of Corolla GLI (2010–2018):")
print(gli_median)

annual_depreciation = (gli_median.loc[2018] - gli_median.loc[2010]) / (2018 - 2010)
print(f"T4f: Overall average value loss per year: {annual_depreciation:.3f} lakh (Rs {annual_depreciation * 100_000:,.0f}/year)")

# (g) The 5 most expensive ads
top5_expensive = df.nlargest(5, "Price")[["Brand", "Model", "Year", "Price", "Price_lakh", "Condition"]]
print("\nT4g: 5 most expensive ads:")
print(top5_expensive)

print("T4g:", df[df["Model"].isin(["Prius","Surf","Land Cruiser"])].groupby(["Model"])["Price_lakh"].median())

# (h) Number of ads and median price for Cash vs Installment/Leasing
trans_summary = df.groupby("Transaction Type")["Price_lakh"].agg(
    Ads_Count="count",
    Median_Price_lakh="median"
)
print("\nT4h: Cash vs Installment/Leasing summary:")
print(trans_summary)
```
